````python
from sklearn.model_selection import train_test_split

from qna_file import SELECTED_FEATURES

filtered_df = df[SELECTED_FEATURES].copy()


import numpy as np

bot_moves = ['move_stockfish_1', 'move_stockfish_9', 'move_stockfish_15',
       'move_maia2_2050', 'move_allie_2500']

filtered_df["matches_engine_move"] = filtered_df[bot_moves].eq(filtered_df["move_player"], axis=0).any(axis=1)

filtered_df.loc[:,"matches_engine_move_encoded"] = filtered_df["matches_engine_move"]*1


filtered_df["player_color_encoded"] = filtered_df['player_color'].apply(lambda x: x == "white")*1




X = filtered_df[['player_color_encoded', 'player_elo', 'opponent_elo', 'half_move', 'move_thinking_time', 'clock_remaining_time', 
                 'normalized_centipawn_loss', 'matches_engine_move_encoded']]
y = filtered_df['is_cheating_move']

X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.2,       # 20% data for testing, 80% for training
    random_state=42,     # Ensures reproducible results
    stratify=y           # Keeps target class distributions balanced
)
````

#### Resolving the issue

Google gave me 2 solutions to this problem. We will need to scale our data and increase number of iterations.

We will use `MinMaxScaling`, making all our values scale from 0-1. This prevents large numbers from having disproportionate impact on our model training.

The default max iterations for sklearn is 100 we will increase it to 1000 using `max_iter`

````python
#### Run this before the next section

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import MinMaxScaler


pipeline = make_pipeline(
    MinMaxScaler(), 
    LogisticRegression(max_iter=1000, random_state=42)
)

model = pipeline.fit(X_train, y_train)
````

#### Model Evaluation

The next stage is to evaluate how well our model performed.

We do this by providing the model with X_test. The accuracy of the model is determined by how close its prediction is from the actual value

````python
#### Run this before answering Q1

from sklearn.metrics import classification_report, confusion_matrix

predictions = model.predict(X_test)

print("--- Classification Report ---")
print(classification_report(y_test, predictions))

print("--- Labeled Confusion Matrix Grid ---")
cm = confusion_matrix(y_test, predictions)

cm_df = pd.DataFrame(
    cm, 
    index=['Actual NO (0)', 'Actual YES (1)'], 
    columns=['Predicted NO (0)', 'Predicted YES (1)']
)

print(cm_df)
````

#### Q5
Now lets try fixing this garbage model.

Sometimes severe class imbalance can cause the model to perform badly. It promotes models to be lazy, they just predict everything as True to get that free 75% accuracy.

Lets fix that by adjusting the balance

We can do that by adding `class_weight='balanced'`

````python
#### Run this before the next question

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import MinMaxScaler


pipeline = make_pipeline(
    MinMaxScaler(), 
    LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
)

model = pipeline.fit(X_train, y_train)


predictions = model.predict(X_test)

print("--- Classification Report ---")
print(classification_report(y_test, predictions))

print("--- Labeled Confusion Matrix Grid ---")
cm = confusion_matrix(y_test, predictions)

cm_df = pd.DataFrame(
    cm, 
    index=['Actual NO (0)', 'Actual YES (1)'], 
    columns=['Predicted NO (0)', 'Predicted YES (1)']
)

print(cm_df)
````
